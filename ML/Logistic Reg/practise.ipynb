```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn.linear_model as lr
```

```python
df=pd.read_excel('data1.xlsx')
df.head()
df.shape
```

```python
fig,ax=plt.subplots(1,2,figsize=(15,5))
sns.scatterplot(data=df,x=df['age'],y=df['have_insurance'],ax=ax[0])
sns.scatterplot(data=df,x=df['income($)'],y=df['have_insurance'],ax=ax[1])
```

```python
#DATA CLEANING 
```

```python
#INCOME
q1=df['income($)'].quantile(0.25)
q3=df['income($)'].quantile(0.75)
iqr=q3-q1 
low=q1-1.5*iqr
high=q3+1.5*iqr
df_cln=df[(df['income($)']>=low)&(df['income($)']<=high)]
df_cln.head()
df_cln.shape
```

```python
#age 
q1=df['age'].quantile(0.25)
q3=df['age'].quantile(0.75)
iqr=q3-q1 
low=q1-1.5*iqr
high=q3+1.5*iqr
df_cln=df_cln[(df_cln['age']>=low)&(df_cln['age']<=high)]
df_cln.shape
```

```python
df.head()
x=df.drop('have_insurance',axis=1)
y=df['have_insurance']
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2) 
```

```python
reg = lr.LogisticRegression()
```

```python
reg.fit(xtrain,ytrain)
reg.score(xtest,ytest)
```
