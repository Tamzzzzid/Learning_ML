```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df=pd.read_csv('Melbourne_housing_FULL.csv')
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=['Price'],axis=0)
```

```python
df.head()
```

```python
df.nunique()
df.drop(['Address','Date','Postcode','Regionname'],axis=1,inplace=True)
```

```python
abs(df.corr(numeric_only=True)['Price']).sort_values(ascending=False)
```

```python
df.head()
```

```python
dummy=pd.get_dummies(df[['Suburb','SellerG','CouncilArea']],drop_first=True)
dummy.head()
```

```python
df=df.drop(['Suburb','SellerG','CouncilArea'],axis=1)
```

```python
df=pd.concat([df,dummy],axis=1)
print(df.shape)
df.head()
```

```python
dummy=pd.get_dummies(df[['Type']],drop_first=True)
df=df.drop(['Type'],axis=1)
df=pd.concat([df,dummy],axis=1)
```

```python
df.isnull().sum()
```

```python
df['Bedroom2']=df['Bedroom2'].fillna((df['Bedroom2'].mean()+df['Bedroom2'].median())/2)
df['Bathroom']=df['Bathroom'].fillna((df['Bathroom'].mean()+df['Bathroom'].median())/2)
```

```python
df=df.dropna(axis=0)
```

```python
dummy=pd.get_dummies(df[['Method']],drop_first=True)
df=df.drop(['Method'],axis=1)
df=pd.concat([df,dummy],axis=1)
df=df.astype(int)
```

```python
corr=abs(df.corr(numeric_only=True)['Price']).sort_values(ascending=False)
```

```python
corr_df = corr.to_frame(name='Correlation_with_Price')
corr_df.head()
```

```python
corr_df = corr.reset_index()
corr_df.columns = ['col', 'cor']
corr_df.head()
```

```python
arr = []
for index, row in corr_df.iterrows():
    if row['cor'] > 0.01:
        arr.append(row['col'])

print(arr)
```

```python
df=df[arr]
df.head()
```

```python
x=df.drop(['Price'],axis=1)
y=df['Price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2)
print(f"Train size = {xtrain.shape}  Test size = {xtest.shape}")
```

```python
from sklearn import linear_model as ln 
from sklearn.model_selection import GridSearchCV
```

```python
l2=ln.Ridge()
l2.fit(xtrain,ytrain)
```

```python
# from sklearn.preprocessing import StandardScaler

# scaler = StandardScaler()
# xtrain = scaler.fit_transform(xtrain)
# xtest = scaler.transform(xtest)
```

```python
l2.score(xtest, ytest)
```

```python
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV


gvc = GridSearchCV(
    Ridge(), {
        'alpha': [0.01, 0.1, 1.0, 10.0, 100.0, 500.0]
    }, cv=5, return_train_score=False
)

gvc.fit(xtrain, ytrain)

# 3. Access results using cv_results_
df = pd.DataFrame(gvc.cv_results_)
df.head(3)

```

```python
df=df[['param_alpha','mean_test_score']].sort_values(by='mean_test_score', ascending=False)
df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
xtrain = scaler.fit_transform(xtrain)
xtest = scaler.transform(xtest)


```

```python
import pandas as pd
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV

gvc = GridSearchCV(
    Lasso(max_iter=10000), {  
        'alpha': [500.0,1000.0,1500.0,2000.0,5000.0]
    }, cv=5, return_train_score=False
)

gvc.fit(xtrain, ytrain)


```
