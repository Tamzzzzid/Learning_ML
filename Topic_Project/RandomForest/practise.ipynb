```python
import pandas as pd
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import accuracy_score, classification_report ,f1_score,roc_auc_score
from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier
import numpy as np
```

```python
df=pd.read_csv('liver_patient_dataset.csv')
df.tail(5)
```

```python
df['Selector']=df['Selector'].replace('Liver Disease',1)
df['Selector']=df['Selector'].replace('No Liver Disease',0)
df['Selector']=df['Selector'].astype(int)
```

```python
df['Gender']=df['Gender'].replace({
    'Male' : 1,
    'Female' : 0
})
df['Gender']=df['Gender'].astype(int)
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8)
```

DECISION TREE - WITHOUT TUNING

```python
dtc=DecisionTreeClassifier()
dtc.fit(xtrain,ytrain)
print(print(f"Accurecy Score {accuracy_score(ytest,dtc.predict(xtest))}  F1 score = {f1_score(ytest,dtc.predict(xtest))}"))
```

DECISION TREE - WITH TUNING

```python
params={
    'criterion' : ['gini','entropy'],
    'max_depth' : [None,5,10,15,20,30,50],
    'min_samples_leaf' : [1,5,10,20,30,50]
}
dt1=DecisionTreeClassifier()
gd1=GridSearchCV(dt1,params,cv=5,n_jobs=-1,scoring='accuracy')
gd1.fit(xtrain,ytrain)
print(print(f"Accurecy Score {accuracy_score(ytest,gd1.predict(xtest))}  F1 score = {f1_score(ytest,gd1.predict(xtest))}"))
```

```python
gd1.best_params_
```

```python
msl = np.linspace(25, 30, 10, dtype=int)
params={
    'criterion' : ['gini'],
    'max_depth' : [5,10,15,20,100],
    'min_samples_leaf' : msl
}
dt1=DecisionTreeClassifier()
gd1=GridSearchCV(dt1,params,cv=5,n_jobs=-1,scoring='accuracy')
gd1.fit(xtrain,ytrain)
print(print(f"Accurecy Score {accuracy_score(ytest,gd1.predict(xtest))}  F1 score = {f1_score(ytest,gd1.predict(xtest))}"))
```

```python
gd1.best_params_
```

RANDOM FOREST 

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf=RandomForestClassifier()
params={
    'criterion' : ['gini'],
    'max_depth' : [10,15,20,50],
    'min_samples_leaf' : [10,20,26,30,40],
    'n_estimators' : [50,100,500],
    'bootstrap' : [True],
    'max_features' : ['sqrt','log2'],
    'max_samples' : [0.7,0.6]
}
gd1=GridSearchCV(rf,params,cv=5,n_jobs=-1,scoring='accuracy')
gd1.fit(xtrain,ytrain)
y_pred = gd1.predict(xtest)
print(f"Accuracy Score = {accuracy_score(ytest, y_pred)}")
print(f"F1 Score = {f1_score(ytest, y_pred)}")
print(f"Best Params = {gd1.best_params_}")
```

```python
rf=RandomForestClassifier()
params={
    'max_depth': [4, 6, 8, 12, None],
    'min_samples_leaf': [5, 10, 20, 30],
    'max_features': ['sqrt', 'log2', 0.5],
    'max_samples': [0.5, 0.7, 0.9],
    'n_estimators': [300],
    'class_weight': [None, 'balanced_subsample']

}
gd1=GridSearchCV(rf,params,cv=5,n_jobs=-1,scoring='roc_auc')
gd1.fit(xtrain,ytrain)
y_pred = gd1.predict(xtest)
proba = gd1.predict_proba(xtest)[:, 1]
print("ROC AUC (proba):   ", roc_auc_score(ytest, proba))
print(f"F1 Score = {f1_score(ytest, y_pred)}")
print(f"Best Params = {gd1.best_params_}")
```
