Cleaning 
->Table Patient

```python
dfp.head(3)
```

```python
#1
#dropping patient who were registered multiple times with different id 
```

```python
dfp.drop_duplicates(subset=['given_name','surname'],keep='first',inplace=True) 
```

```python
#2
#dropping rows where 'address','city','state','zip_code','country','contact' have null values
```

```python
dfp.dropna(subset=['address','city','state','zip_code','country','contact'],inplace=True)
```

```python
#3 invalid zip codes are converted to valid code
```

```python
dfp['zip_code']=dfp['zip_code'].astype('Int64').astype(str)
valid_Ind=dfp['zip_code'] != '<NA>'
dfp.loc[valid_Ind,'zip_code']=dfp.loc[valid_Ind,'zip_code'].str.zfill(5)
```

```python
#4 dropping dwarf XD
dfp.drop(dfp[dfp['height']<=30.00].index,inplace=True)
dfp.drop(dfp[dfp['weight']<=50.00].index,inplace=True)
```

```python
dfp.describe()
```

```python
dfp[['contact']]
```

```python

#4 fixing email 
phone_pattern = r'((?:\+?1\s?)?\(?\d{3}\)?[\s.-]?\d{3}[\s.-]?\d{4})'


dfp['phone_number'] = dfp['contact'].str.extract(phone_pattern)
dfp['email'] = dfp.apply(lambda row: row['contact'].replace(str(row['phone_number']), '').strip('+ '), axis=1)
dfp = dfp.drop(columns=['contact'])
dfp.head()

```

```python
display(dft.head(3))
display(dfc.head(3))
print(dft.shape)
```

```python
#adding dft and dfc
dft=pd.concat([dft,dfc],ignore_index=True)
dft.shape
```

```python
#removing u
dft['novodra']=dft['novodra'].str.replace('u','')
dft['auralin']=dft['auralin'].str.replace('u','')
```

```python
dft.head(3)
```

```python
#ccombining auralin and novodra 
aur_val = dft['auralin'] != '-'
nov_val = dft['novodra'] != '-'

dft.loc[aur_val, 'range'] = dft.loc[aur_val, 'auralin']
dft.loc[aur_val, 'type'] = 'auralin'

dft.loc[nov_val, 'range'] = dft.loc[nov_val, 'novodra']
dft.loc[nov_val, 'type'] = 'novodra'

dft.drop(columns=['auralin', 'novodra'], inplace=True)

dft.head(3)
```

```python
#spliting range 
dft[['before_drug','after_drug']]=dft['range'].str.split('-',expand=True).astype(int)
```

```python
dft.drop(['range'],inplace=True,axis=1)
dft.head(3)
```

```python
#fixing hba1c_change 
dft['hba1c_change']=dft['hba1c_start']-dft['hba1c_end'] 
dft.head(3)
```

```python
#marging dft and dfp
```

```python
dfp['given_name'] = dfp['given_name'].str.lower()
dfp['surname'] = dfp['surname'].str.lower()
df=pd.merge(
    dft,
    dfp[['patient_id','given_name','surname']],
    on=['given_name','surname'],
    how='left'
)
df.head(5)
```

```python
df.dropna(inplace=True)
```
