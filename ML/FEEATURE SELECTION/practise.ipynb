```python
import pandas as pd 
import numpy as np
```

```python
df=pd.read_excel('ecommerce_dataset_100x100.xlsx')
df.head(3)
```

```python
arr=df.columns
num=[]
cat=[]
```

```python
def separetor (arr) : 
    for col in arr :
        converted = pd.to_numeric(df[col], errors='coerce') 
        if ((df[col].nunique()/ df.shape[0] )*100) > 30 : 
            if converted.notna().sum() > 0 and (converted.nunique() / df.shape[0]) * 100 > 30:
                num.append(col)    
        else : 
            cat.append(col) 


```

```python
separetor(arr)
print(f"num = {len(num)} cat = {len(cat)}")
dfn=df[num]
dfc=df[cat]
dfn.head()
```

```python
y = dfc['ReturnRequested'].map({'Yes': 1, 'No': 0, 'True': 1, 'False': 0})
y = pd.to_numeric(y, errors='coerce').astype('Int64')
dfc.drop(['ReturnRequested'], axis=1, inplace=True)
```

```python
dfn.drop(['ActualDeliveryDate'],axis=1,inplace=True)
dfn.head(3)
```

```python

dfn['OrderMonth'] = dfn['OrderDate'].dt.month
dfn['Orderday'] = dfn['OrderDate'].dt.day

dfn['DeliveryMonth'] = dfn['EstimatedDeliveryDate'].dt.month
dfn['Deliveryday'] = dfn['EstimatedDeliveryDate'].dt.day

dfn['SignupMonth'] = dfn['SignupDate'].dt.month
dfn['Signupday'] = dfn['SignupDate'].dt.day

dfn['BirthYear'] = dfn['DateOfBirth'].dt.year

dfn = dfn.drop(columns=['OrderDate', 'EstimatedDeliveryDate', 'SignupDate', 'DateOfBirth'])
```

```python
dfn.drop(['LastLoginDate'],axis=1,inplace=True)
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
scaled_data = sc.fit_transform(dfn)
dfn = pd.DataFrame(scaled_data, columns=dfn.columns, index=dfn.index)
dfn.head(3)




```

```python
dfc=pd.get_dummies(dfc,drop_first=True)
dfc.head()
```

```python

dfc.drop(['ReturnDate','DeliveryDays'],axis=1,inplace=True)
info_df = pd.DataFrame({

    'Column': dfc.columns,

    'Non-Null Count': dfc.notna().sum().values,

    'Dtype': dfc.dtypes.astype(str).values,

    'Null Count': dfc.isna().sum().values

})

info_df.head(3)
```

```python
bool_cols = dfc.select_dtypes(include='bool').columns
dfc[bool_cols] = dfc[bool_cols].astype(int)
dfc.head(3)
```

```python
df=pd.concat([dfn,dfc],axis=1)
scaled_data = sc.fit_transform(df)
df = pd.DataFrame(scaled_data, columns=df.columns, index=df.index)
df.head(3)
```
