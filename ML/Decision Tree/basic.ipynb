```python
%matplotlib inline
import pandas as pd
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.linear_model as lr
import numpy as np
```

```python
df=pd.read_excel('salary.xlsx')
df.head()
```

```python
fg,ax=plt.subplots(1,3,figsize=(20,4)) 
sns.scatterplot(data=df,x='company',y='salary_more_then_100k',ax=ax[0])
sns.scatterplot(data=df,x='job',y='salary_more_then_100k',ax=ax[1])
sns.scatterplot(data=df,x='degree',y='salary_more_then_100k',ax=ax[2])
plt.tight_layout()
```

```python
x=df.drop('salary_more_then_100k',axis=1)
y=df['salary_more_then_100k']
```

```python
from sklearn.preprocessing import LabelEncoder 
le=LabelEncoder()
x=x.apply(le.fit_transform)
x.head()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2)
```

```python
from sklearn import tree 
reg=tree.DecisionTreeClassifier()
reg.fit(xtrain,ytrain)
acc=reg.score(xtest,ytest)
print(f"Accuracy of the model is: {acc*100:.2f}%")
```
