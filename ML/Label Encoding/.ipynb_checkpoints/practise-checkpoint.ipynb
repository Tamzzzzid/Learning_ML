```python
"""
Nominal Data
Nominal data consists of categories that have no inherent order or ranking. They are simply distinct labels or names. 
You cannot logically sort them from highest to lowest or best to worst.
•	Examples:
•	Eye color (Brown, Blue, Green)
•	Blood types (A, B, AB, O)
•	Types of pets (Dog, Cat, Bird)
Ordinal Data
Ordinal data consists of categories that do have a natural, logical order or ranking. 
However, the exact distance or measurable difference between each category isn't known or isn't necessarily equal.
•	Examples:
•	Customer satisfaction ratings (Poor, Fair, Good, Excellent)
•	Education levels (High School, Bachelor's, Master's, PhD)
•	Spiciness levels on a menu (Mild, Medium, Hot, Extra Hot)
"""
```
