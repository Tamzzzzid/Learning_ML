```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.datasets import make_classification
```

```python
x, y = make_classification(n_samples=1000, n_features=20, random_state=42)
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
base=DecisionTreeClassifier()
```

```python
pip=Pipeline([
    ('scale',StandardScaler()),('estimator',AdaBoostClassifier(estimator=base))
])
```

```python
param={
    'estimator__n_estimators' : [50, 100, 200, 500],
    'estimator__learning_rate' : [0.01, 0.1, 0.5, 1.0],
    'estimator__estimator__max_depth' : [1,2]

}
```

```python
gv=GridSearchCV(pip,param,cv=5,scoring='accuracy',n_jobs=-1)
gv.fit(xtrain,ytrain)
```

```python
y_pred = gv.predict(xtest)
print(gv.best_params_)
print(f"\nTest Set Accuracy: {accuracy_score(ytest, y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(ytest, y_pred))
```
