```python
import pandas as pd
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import accuracy_score, classification_report
from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier
import numpy as np
```

```python
data = load_breast_cancer()
x = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target)
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8)
```

```python
param={
    'criterion' : ['gini','entropy'],
    'max_depth' :[None,5,10,15,20],
    'min_samples_split' : [4,10,12,20],
    'min_samples_leaf' : [2,8,20],
    'max_features' : ['sqrt','log2',None]
}
```

```python
dt=DecisionTreeClassifier()
dt.fit(xtrain,ytrain)
print(accuracy_score(ytest,dt.predict(xtest)))
```

```python
dtg=DecisionTreeClassifier()
gd=GridSearchCV(
    dtg,param,cv=5,scoring='accuracy',n_jobs=-1

)
gd.fit(xtrain,ytrain)
print(accuracy_score(ytest,gd.predict(xtest)))
```

```python
gd.best_params_
```
