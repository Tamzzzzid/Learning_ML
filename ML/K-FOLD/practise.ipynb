```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.model_selection import train_test_split,cross_val_score 
from sklearn.ensemble import RandomForestClassifier 
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_excel("heart_ready.xlsx")
df.head()
```

```python
x=df.drop('HeartDisease',axis=1)
y=df['HeartDisease']
x.head()
```

```python
ar=cross_val_score(LogisticRegression(max_iter=10000),x,y,cv=5)
print((ar.sum()/5)*100)
```

```python
arr=cross_val_score(RandomForestClassifier(n_estimators=80,max_features=4,max_depth=10),x,y,cv=10)
print((arr.sum()/10)*100)
```
