```python
import pandas as pd 
import numpy as np
```

```python
df=pd.read_csv('carprices.csv')
df
```

```python
y=df['Sell Price($)']
cat=df[['Car Model']]
num=df.drop(['Car Model','Sell Price($)'],axis=1)
```

```python
from sklearn.preprocessing import OneHotEncoder as oh
```

```python
ohe=oh(sparse_output=False,drop='first')
cat_data=ohe.fit_transform(cat) 
x=pd.DataFrame(cat_data,columns=ohe.get_feature_names_out())
x=pd.concat([x,num],axis=1)
x
```

```python
import sklearn.linear_model as lr 
reg=lr.LinearRegression()
```

```python
reg.fit(x,y)
reg.score(x,y)*100
```

```python
reg.predict([[0,0,60000,4]])
```
