```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_blobs
```

```python
x,y=make_blobs(n_samples=1000,n_features=2,centers=3,random_state=43)
```

```python
sns.scatterplot(x=x[:,0],y=x[:,1])
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y)
```

```python
from sklearn.cluster import KMeans
```

```python
arr=[]
for k in range(1,12) : 
    km=KMeans(n_clusters=k,init='k-means++')
    km.fit(xtrain)
    arr.append(km.inertia_)
print(arr)
```

```python
sns.lineplot(x=range(1, 12,1), y=arr, marker='o')
```

```python
km=KMeans(n_clusters=3,init='k-means++')
km.fit(xtrain)
y=km.predict(xtrain)
sns.scatterplot(x=xtrain[:,0],y=xtrain[:,1],c=y)
```

```python
sns.scatterplot(x=xtest[:,0],y=xtest[:,1],c=km.predict(xtest))
```
