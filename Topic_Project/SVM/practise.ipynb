```python
from sklearn.datasets import load_breast_cancer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.svm import SVC

import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('train_and_test2.csv')
df=df.dropna()
```

```python
df.rename(columns={
    '2urvived': 'survived'
}, inplace=True)

x=df.drop(['survived'],axis=1)
y=df['survived']
df.head(3)
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8)
```

```python
pip=Pipeline(
    [('scale',StandardScaler()),
    ('svm',SVC())]
)
```

```python
param = [
    {
        'svm__kernel': ['linear'],
        'svm__C': [0.01,0.1, 1, 10]
    },
    {
        'svm__kernel': ['rbf'],
        'svm__C': [0.01,0.1, 1, 10],
        'svm__gamma': ['scale',0.01, 0.1,1]
    },
    {
        'svm__kernel': ['poly'],
        'svm__C': [0.01,0.1, 1, 10],
        'svm__gamma': ['scale'],
        'svm__degree': [2, 3]
    },
    {
        'svm__kernel': ['sigmoid'],
        'svm__C': [0.01,0.1, 1, 10],
        'svm__gamma': ['scale',0.1,1]
    }
]
```

```python
mod=GridSearchCV(pip,param,cv=5,scoring='accuracy')
```

```python
mod.fit(xtrain,ytrain)
```

```python
test_accuracy = mod.score(xtest, ytest)
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
df=mod.cv_results_
df=pd.DataFrame(df).sort_values(by='mean_test_score',ascending=False)
#df[['param_svm__kernel','mean_test_score','params']].head(10)
df.head(10)
```
