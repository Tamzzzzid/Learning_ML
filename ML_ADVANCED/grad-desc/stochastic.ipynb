```python
from sklearn.datasets import load_diabetes 
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
x,y=load_diabetes(return_X_y=True)
print(f" X = {x.shape}  Y = {y.shape} ")
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lnreg=LinearRegression()
lnreg.fit(xtrain,ytrain) 
ypred=lnreg.predict(xtest)   
print(f"Coef = {lnreg.coef_} \n interceptor = {lnreg.intercept_} \n R2 score = {r2_score(ytest,ypred)}")
```

```python
class Stoc_grad : 
    def __init__(self,learning_rate=0.01,iter=100) :
        self.coef= None
        self.interceptor= None 
        self.learning_rate=learning_rate
        self.iter=iter 
    
    def fit(self,xtrain,ytrain) :
        self.interceptor=0
        self.coef=np.ones(xtrain.shape[1])
        for i in range(self.iter) :
            for j in range(xtrain.shape[0]) : 
                ix=np.random.randint(0,xtrain.shape[0])

                y_hat=np.dot(xtrain[ix],self.coef)+self.interceptor
                def_interceptor=-2 * (ytrain[ix] - y_hat)
                self.interceptor=self.interceptor - (self.learning_rate * def_interceptor)

                def_coef=-2 * np.dot((ytrain[ix] - y_hat),xtrain[ix])
                self.coef=self.coef - (self.learning_rate * def_coef)


    def predict(self,xtest) : 
        return np.dot(xtest,self.coef)+self.interceptor
```

```python
reg=Stoc_grad(0.05,100)
reg.fit(xtrain,ytrain)
pred=reg.predict(xtest)
r2=r2_score(ytest,pred)
print(r2)
```

```python
arr=[]
learning_rate= [0.001, 0.01,0.05 ,0.1,0.5,0.7,0.9]
iter = [100,200,400,800,1000,2000]
for it in iter : 
    for lr in learning_rate :
        reg=Stoc_grad(lr,it)
        reg.fit(xtrain,ytrain)
        pred=reg.predict(xtest)
        r2=r2_score(ytest,pred)
        arr.append([it,lr,r2])
df=pd.DataFrame(arr,columns=['Iteration','learning_rate','R2_score']).sort_values(by='R2_score',ascending=False)    
display(df.head(5))
```
