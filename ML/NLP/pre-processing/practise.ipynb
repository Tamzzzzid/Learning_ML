#2 Remove html tags 

```python
import re
```

```python
def html_remover(txt) : 
    pattern=re.compile('<.*?>')
    return pattern.sub(r'',txt)
```

```python
txt=df['review'][3]
print(txt)
txt=html_remover(txt)
print(txt)
```

```python
print(df['review'][2])
df['review']=df['review'].apply(html_remover)
print(df['review'][2])
```

#3 remove url

```python
def url_remover(txt) : 
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',txt)
```

```python
print(df['review'][2])
df['review']=df['review'].apply(url_remover)
print(df['review'][2])
```

```python
df['review']=df['review'].str.replace('(see:','')
print(df['review'][2])
```

#4 remove punctuation

```python
import string,time 
exclude=string.punctuation
```

```python
#option 1 (very slow approach)
def removr_punc(txt) : 
    for i in exclude : 
        txt=txt.replace(i,'')
    return txt
```

```python
start=time.time()
txt=df['review'][2]
print(txt)
txt=removr_punc(df['review'][2])
print(txt)
end=start-time.time()
print(end)
```

```python
#option 2 (faster approach)
def removr_punc2(txt) :
    return txt.translate(str.maketrans('','',exclude))
```

```python
start=time.time()
txt=df['review'][2]
print(txt)
txt=removr_punc(df['review'][2])
print(txt)
end2=start-time.time()
print(end2)
```

```python
#how faster? 
print(end/end2)
```

```python
print(df['review'][2])
df['review']=df['review'].apply(removr_punc2)
print(df['review'][2])
```

```python
def slang_remover(txt) : 
    if not isinstance(txt,str) :
        return txt
    words=txt.split()
    converted=[]
    for word in words :
        word=word.upper() 
        if word in slangs :
            converted.append(slangs[word])
        else : 
            converted.append(word)
    return ' '.join(converted)
```
